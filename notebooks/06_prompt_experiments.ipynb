# Notebook 06: Prompt Engineering Experiments for Support Reply Drafting
SupportIQ - Syllabus Experiment 4

Compares three prompt strategies:
1. **Zero-Shot Prompting**: Direct instruction + ticket signals.
2. **Few-Shot Prompting**: Demonstrating tone and style through 3 reviewed examples.
3. **Structured Prompting**: Bounding input fields with strict format constraints.

Evaluated via Human Rubric: Relevance, Correctness, Professional Tone, Empathy, Completeness, Hallucination Check.

```python
import sys
sys.path.append('../ml_service')
from app.reply_generation.generator import ReplyGenerator

gen = ReplyGenerator()
ticket = 'My debit card was blocked after 3 incorrect PIN entries. I need to pay for groceries now!'
intent = 'pin_blocked'
sentiment = 'negative'
priority = 'HIGH'
summary = 'Customer card blocked due to 3 incorrect PIN entries with urgent purchase need.'

for p_type in ['ZERO_SHOT', 'FEW_SHOT', 'STRUCTURED']:
    res = gen.generate(ticket, intent, sentiment, priority, summary, prompt_type=p_type)
    print(f'=== Strategy: {p_type} ===')
    print(res['text'])
    print('-'*50)
```
