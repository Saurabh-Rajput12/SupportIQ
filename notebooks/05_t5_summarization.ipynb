Demonstrates conditional sequence-to-sequence generation using `google-t5/t5-small` to produce concise 1-2 sentence ticket summaries for agent review.

```python
import sys
sys.path.append('../ml_service')
from app.summarization.summarizer import T5Summarizer
import json

summarizer = T5Summarizer()
long_ticket = '''
Dear Support, I checked my bank account balance this morning and noticed that an international wire transfer of 500 Euros was charged twice on October 30th. Both entries have the identical transaction ID TX-88291. I only authorized one transfer to my landlord. Please review the duplicate charge and refund the second 500 Euros immediately as my rent is due.
'''

summary = summarizer.summarize(long_ticket)
print('T5 Summary Result:')
print(json.dumps(summary, indent=2))
```
