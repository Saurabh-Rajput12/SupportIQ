# Notebook 01: Dataset Exploration & EDA (BANKING77)
SupportIQ - MCA Data Science Experiment 1 (Data Exploration Baseline)

This notebook inspects the PolyAI BANKING77 dataset, analyzes class distribution, checks for nulls/duplicates, and measures sequence length distributions.

```python
from datasets import load_from_disk
import pandas as pd
import json

ds = load_from_disk('../data/raw/banking77')
train_df = pd.DataFrame(ds['train'])
test_df = pd.DataFrame(ds['test'])
print(f'Train shape: {train_df.shape}, Test shape: {test_df.shape}')
```

```python
with open('../data/processed/banking77_id2label.json', 'r') as f:
    id2label = json.load(f)

train_df['intent_name'] = train_df['label'].map(lambda x: id2label.get(str(x), str(x)))
train_df.head()
```

```python
train_df['char_length'] = train_df['text'].str.len()
train_df['word_count'] = train_df['text'].str.split().str.len()
train_df[['char_length', 'word_count']].describe()
```

```python
class_counts = train_df['intent_name'].value_counts()
print('Top 5 intents:')
print(class_counts.head())
print('\nBottom 5 intents:')
print(class_counts.tail())
```
