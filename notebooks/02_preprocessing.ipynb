# Notebook 02: NLP Preprocessing & Tokenization Pipeline
SupportIQ - Syllabus Experiment 1

Demonstrates the preprocessing policy:
1. Text cleaning and whitespace normalization without stripping essential semantic cues.
2. Preservation of punctuation, negations ('not', 'never'), and contractions.
3. Tokenization using `google-bert/bert-base-uncased` with attention masks and max-sequence truncation.

```python
import sys
sys.path.append('../ml_service')
from app.preprocessing.cleaner import TextCleaner
from app.preprocessing.tokenizer import TokenizerService

raw_ticket = '''
Hello SupportIQ,

I did NOT authorize this transaction of $340.50 on my card yesterday!! 
The merchant was 'Acme Electronics' but I never shopped there. 
Please block my card immediately and refund the amount.
'''

cleaned_text = TextCleaner.clean(raw_ticket)
print('Cleaned Text:\n', cleaned_text)
```

```python
tokenizer_svc = TokenizerService()
tokens = tokenizer_svc.tokenize(cleaned_text, max_length=64)
print('Token Count:', tokens['token_count'])
print('Tokens:', tokens['tokens'][:20])
print('Input IDs:', tokens['input_ids'][:10])
print('Attention Mask:', tokens['attention_mask'][:10])
```
