# Notebook 04: Sentiment Analysis & Priority Recommendation Engine
SupportIQ - Syllabus Supporting Experiment

Analyzes customer support tone using DistilBERT SST-2 (binary classification: positive / negative) and runs the explainable rule-based priority engine (CRITICAL, HIGH, MEDIUM, LOW).

```python
import sys
sys.path.append('../ml_service')
from app.sentiment.analyzer import SentimentAnalyzer
from app.priority.engine import PriorityEngine
import json

sentiment_analyzer = SentimentAnalyzer()
sample_ticket = 'My wallet was stolen with my credit card inside! Someone is trying to withdraw cash right now!'
sent_res = sentiment_analyzer.predict(sample_ticket)
print('Sentiment:', json.dumps(sent_res, indent=2))
```

```python
priority_res = PriorityEngine.evaluate(sample_ticket, intent_label='lost_or_stolen_card', sentiment=sent_res)
print('Priority Recommendation:', json.dumps(priority_res, indent=2))
```
